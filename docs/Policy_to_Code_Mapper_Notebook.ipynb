## 1. Executive Summary & Research Methodology

Security, privacy, and compliance policies are frequently disconnected from daily software development practices, buried in intranet wikis or static PDFs. Developers rarely consult these repositories while writing features, causing policy awareness to arrive too late during audits or security incidents.

**Policy-to-Code Mapper** brings policy guidance directly into Visual Studio Code at the point of coding using human-centered behavioral nudges:

> **Code analysis is the trigger. Policy guidance is the product. Developer understanding is the desired outcome.**

## 3. Local Code Context Detection Engine
Test deterministic local pattern matching on Python code snippets.

```python
import re

def analyze_python_snippet(code_snippet, policies):
    findings = []
    lines = code_snippet.strip().split('\n')
    
    for line_idx, line_text in enumerate(lines, 1):
        line_text_trimmed = line_text.strip()
        if not line_text_trimmed or line_text_trimmed.startswith('#'):
            continue
            
        for p in policies:
            # Check logging function matches
            for fn in p.get('loggingFunctions', []):
                if fn in line_text_trimmed:
                    for ident in p.get('sensitiveIdentifiers', []):
                        pattern = r'(?<![a-zA-Z0-9_])' + re.escape(ident) + r'(?![a-zA-Z0-9_])'
                        if re.search(pattern, line_text_trimmed, re.IGNORECASE):
                            findings.append({
                                'line': line_idx,
                                'code': line_text_trimmed,
                                'policy_id': p['id'],
                                'policy_title': p['title'],
                                'matched_identifier': ident,
                                'severity': p['severity']
                            })
                            
            # Check OWASP SQL Injection
            if p['id'] == 'SEC-INJ-001' and 'execute' in line_text_trimmed:
                if re.search(r'execute\s*\(', line_text_trimmed):
                    findings.append({
                        'line': line_idx,
                        'code': line_text_trimmed,
                        'policy_id': p['id'],
                        'policy_title': p['title'],
                        'matched_identifier': 'unsanitized_sql',
                        'severity': p['severity']
                    })
                    
            # Check OWASP Hardcoded Secrets
            if p['id'] == 'SEC-CRY-001':
                if re.search(r'(api_key|secret_key|access_token)\s*=\s*["\'][a-zA-Z0-9_\-\.]{8,}["\']', line_text_trimmed, re.IGNORECASE):
                    findings.append({
                        'line': line_idx,
                        'code': line_text_trimmed,
                        'policy_id': p['id'],
                        'policy_title': p['title'],
                        'matched_identifier': 'hardcoded_secret',
                        'severity': p['severity']
                    })
                    
    return findings

# Sample Python test code
sample_code = """
api_key = "sk_live_1234567890abcdef12345"
print("User password:", password)
logger.info("Login attempt for %s with password %s", username, password)
logger.info("User email: %s", email)
cursor.execute(f"SELECT * FROM users WHERE username = '{username}'")
logger.info("Application server initialized on port 8080")
"""

findings = analyze_python_snippet(sample_code, policies)
print(f"Analysis detected {len(findings)} policy considerations:\n")
for f in findings:
    print(f"[Line {f['line']}] [{f['severity'].upper()}] {f['policy_id']} — {f['policy_title']}")
    print(f"  Code: {f['code']}")
    print(f"  Matched Pattern: {f['matched_identifier']}\n")
```

## 4. Policy Traceability Chain Visualizer
Demonstrates the 4-step provenance from source code to regulatory references and safer implementation examples.

```python
def render_traceability_chain(finding, policies):
    p = next((pol for pol in policies if pol['id'] == finding['policy_id']), None)
    if not p:
        return
        
    print("=" * 60)
    print(f"🛡️ POLICY-TO-CODE TRACEABILITY CHAIN — {p['id']}")
    print("=" * 60)
    print(f"1. Source Code Line:\n   {finding['code']}\n")
    print(f"2. Observed Code Pattern:\n   Identifier '{finding['matched_identifier']}' in monitored function\n")
    print(f"3. Matched Policy Requirement:\n   {p['id']}: {p['title']}\n   {p['description']}\n")
    
    if p.get('supportingRegulatoryContext'):
        reg = p['supportingRegulatoryContext'][0]
        print(f"⚖️ Supporting Regulatory Context:\n   {reg['framework']} {reg['article']} ({reg['title']}): {reg['relationship']}\n")
        
    print(f"4. Suggested Developer Action:\n   {p['suggestedAction']}")
    print(f"   Safer Example:\n   {p['saferExample']}\n")
    print("Disclaimer: This is automated developer guidance, not legal advice or a legal-compliance determination.")
    print("=" * 60 + "\n")

if findings:
    render_traceability_chain(findings[0], policies)
```

### Precision & Recall Formulae:
$$\text{Precision} = \frac{TP}{TP + FP}, \quad \text{Recall} = \frac{TP}{TP + FN}$$

Where $TP$ = True Positives, $FP$ = False Positives, $FN$ = False Negatives.
